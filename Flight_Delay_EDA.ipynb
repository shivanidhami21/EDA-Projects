```python
plt.figure(figsize=(8,5))
sns.histplot(df["Arrival_Delay"], bins=30, kde=True)
plt.title("Distribution of Arrival Delay")
plt.xlabel("Arrival Delay")
plt.ylabel("Frequency")
plt.show()
```

Most flights have little or no arrival delay. The distribution is positively skewed, as a few flights have very high arrival delays. Some extreme values may be outliers.

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df["Arrival_Delay"])
plt.title("Boxplot of Arrival Delay")
plt.show()
```

The boxplot shows that the Arrival_Delay column contains many outliers. Most flights have low arrival delays, but a few flights have very high arrival delays.

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Dep_Delay"], bins=30, kde=True)
plt.title("Distribution of Departure Delay")
plt.xlabel("Departure Delay")
plt.ylabel("Frequency")
plt.show()
```

Most flights have low departure delays. A few flights have very high departure delays, so the data is right-skewed.

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df["Dep_Delay"])
plt.title("Boxplot of Departure Delay")
plt.show()
```

The Dep_Delay column contains many outliers. Most flights have low departure delays, while a few flights have very high departure delays.

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="Dep_Delay", y="Arrival_Delay", data=df)
plt.title("Departure Delay vs Arrival Delay")
plt.xlabel("Departure Delay")
plt.ylabel("Arrival Delay")
plt.show()
```

There is a positive relationship between Dep_Delay and Arrival_Delay. As departure delay increases, arrival delay also tends to increase.

```python
plt.figure(figsize=(8,5))
sns.countplot(x="Origin_Airport", data=df)
plt.title("Number of Flights from Each Origin Airport")
plt.xlabel("Origin Airport")
plt.ylabel("Number of Flights")
plt.show()
```

LAX has the highest number of flights. DFW and JFK also have many flights. Some airports have a much lower number of flights compared to others.

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.select_dtypes(include=["number"]).corr(),
            annot=True,
            cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
delay_reasons = df[
    ["Carrier_Delay",
     "WeatherDelay",
     "NAS_Delay",
     "Security_Delay",
     "Late_Aircraft_Delay"]
].sum()

delay_reasons
```

```python
plt.figure(figsize=(8,5))

delay_reasons.plot(kind="bar")

plt.title("Total Delay by Reason")
plt.xlabel("Delay Reason")
plt.ylabel("Total Delay")
plt.xticks(rotation=45)

plt.show()
```

Late Aircraft Delay is the biggest reason for flight delays. Carrier Delay and NAS Delay also contribute significantly. Weather Delay has a smaller impact, while Security Delay is the least common reason for flight delays.
